## Problem 2

### Part (a)
Best choice: Containers

I can package the portal/app with all its dependencies and run the same image for many local sites, just swapping configs (env vars, database URLs, storage). That keeps my code identical across sites and makes updates easy—I push a new image and everyone pulls it—and it works well on any cloud with internet access and plenty of storage. VMs would also work but are heavy (full OS images, slower, more ops). Virtual environments only isolate language packages and can’t capture the system tools/services a full web stack needs. The global environment isn’t reproducible and invites version conflicts.

### Part (b)
Best choice: virtual environment

This is a one-off data-cleaning/report job: I just need Python/R packages isolated so versions don’t clash with other projects, and I can lock exact deps to make the analysis reproducible. Containers would work but are overkill here since I’m not deploying a service or sharing a runtime image. VMs are much heavier than needed. The global environment could do it, but it risks dependency conflicts and “it worked on my machine” issues; a simple venv/Conda keeps things clean without extra ops.

### Part (c)
Best choice: Containers

I can creat a Linux image that includes Python plus the exact open-source imaging tools the model needs, so anyone on the team can run the same setup on their laptop or in the cloud. This gives reproducibility and easy updates without the overhead of a full VM. VMs would also solve the Ubuntu requirement, but they’re heavier to build, share, and boot for each user. Virtual environments only handle Python packages and can’t install/lock the required system imaging tools. The global environment is brittle and likely to break across different machines and OS versions.

### Part (d)
Best choice: virtual environment

I’d use a virtual environment and pin Python 3 so this project always runs with its own interpreter even if my system later installs Python 4 with Swedish commands. That’s the lightest, most direct fix. A **container** could also lock Python 3, but that’s heavier than I need for local work. A VM is massive overkill. The global environment is the worst choice—once the system upgrades, my code could break.
